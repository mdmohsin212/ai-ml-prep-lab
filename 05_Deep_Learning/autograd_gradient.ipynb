```python
import torch
```

```python
x = torch.tensor(3.0, requires_grad=True)

print("x:", x)
print("Requires Grad:", x.requires_grad)
```

```python
loss = x**2 + 2*x + 1

print("Loss:", loss)
```

```python
loss.backward()

print("Gradient of loss with respect to x:", x.grad)
```

### Multiple variable

```python
a = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(4.0, requires_grad=True)

loss = a**2 + b**2 + 3*a*b
print("Loss:", loss)
```

```python
loss.backward()

print("Gradient with respect to a:", a.grad)
print("Gradient with respect to b:", b.grad)
```

### gradient reset

```python
a.grad.zero_()
b.grad.zero_()

print("Gradient of a after reset:", a.grad)
print("Gradient of b after reset:", b.grad)
```
