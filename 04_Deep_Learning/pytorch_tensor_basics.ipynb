```python
import torch
```

### PyTorch tensor

```python
tensor_1 = torch.tensor([1, 2, 3, 4, 5])

print("Tensor 1:", tensor_1)
print("Shape:", tensor_1.shape)
print("Data type:", tensor_1.dtype)
```

```python
tensor_2 = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print("Tensor 2:")
print(tensor_2)
print("Shape:", tensor_2.shape)
```

```python
random_tensor = torch.rand(3, 4)

print("Random Tensor:")
print(random_tensor)
print("Shape:", random_tensor.shape)
```

```python
zeros_tensor = torch.zeros(2, 3)
ones_tensor = torch.ones(2, 3)

print("Zeros Tensor:")
print(zeros_tensor)

print("Ones Tensor:")
print(ones_tensor)
```

### Tensor reshape

```python
x = torch.arange(1, 13)

print("Original Tensor:")
print(x)
print("Original Shape:", x.shape)
```

```python
x_reshaped = x.reshape(3, 4)

print("Reshaped Tensor:")
print(x_reshaped)
print("New Shape:", x_reshaped.shape)
```

```python
x_reshaped_2 = x.view(2, 6)

print("Another Reshaped Tensor:")
print(x_reshaped_2)
print("Shape:", x_reshaped_2.shape)
```

### Device transfer

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Available Device:", device)
```

```python
x_device =  x_reshaped.to(device)

print("Tensor Device:", x_device.device)
print(x_device)
```

```python
x_cpu = x_device.to("cpu")

print("Back to CPU:")
print(x_cpu)
print("Device:", x_cpu.device)
```

### Basic math operation

```python
a = torch.tensor([10, 20, 30])
b = torch.tensor([1, 2, 3])

print("Tensor a:", a)
print("Tensor b:", b)
```

```python
addition = a + b
print("Addition:", addition)
```

```python
subtraction = a - b
print("Subtraction:", subtraction)
```

```python
multiplication = a * b
print("Multiplication:", multiplication)
```

```python
division = a / b
print("Division:", division)
```

```python
power = a ** 2
print("Power:", power)
```

### Matrix multiplication

```python
matrix_1 = torch.tensor([
    [1, 2],
    [3, 4]
])

matrix_2 = torch.tensor([
    [5, 6],
    [7, 8]
])
```

```python
matmul_result = torch.matmul(matrix_1, matrix_2)

print("Matrix 1:")
print(matrix_1)

print("Matrix 2:")
print(matrix_2)

print("Matrix Multiplication Result:")
print(matmul_result)
```

### Mean, sum, max, min

```python
values = torch.tensor([10.0, 20.0, 30.0, 40.0])

print("Values:", values)

print("Sum:", torch.sum(values))
print("Mean:", torch.mean(values))
print("Max:", torch.max(values))
print("Min:", torch.min(values))
```
