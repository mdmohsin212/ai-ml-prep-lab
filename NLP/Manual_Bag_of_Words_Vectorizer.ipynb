```python
import re
import numpy as np
import pandas as pd
```

```python
corpus = [
    "I love NLP",
    "I love machine learning",
    "NLP is fun"
]

for i, document in enumerate(corpus):
    print(f"Document {i}: {document}")
```

#### Text Preprocessing Function

```python
def preprocess_text(text):
    text = text.lower()
    tokens = re.findall(r"[a-z]+", text)
    return tokens

for document in corpus:
    print(preprocess_text(document))
```

#### Create Manually Vocabulary

```python
unique_words = set()

for document in corpus:
    tokens = preprocess_text(document)
    
    for word in tokens:
        unique_words.add(word)
        
vocabulary = sorted(unique_words)
print("Vocabulary: ", vocabulary)
print("Vocabulary size:", len(vocabulary))
```

```python
word_to_index = {}

for i, word in enumerate(vocabulary):
    word_to_index[word] = i
    
print("Word to index mapping:")
for word, index in word_to_index.items():
    print(f"{word}: {index}")
```

#### Manual Bag of Words Matrix

```python
bow_matrix = np.zeros(
    (len(corpus), len(vocabulary)),
    dtype=int
)

for document_index, document in enumerate(corpus):
    tokens = preprocess_text(document)
    
    for word in tokens:
        if word in word_to_index:
            word_index = word_to_index[word]
            bow_matrix[document_index][word_index] += 1

print(bow_matrix)
```
