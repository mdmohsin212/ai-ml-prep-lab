```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [1.0, 1.0],
    [1.5, 2.0],
    [2.0, 1.0],
    [1.0, 2.0],

    [5.0, 5.0],
    [6.0, 5.0],
    [5.5, 6.0],
    [6.0, 6.0],

    [8.0, 1.0],
    [9.0, 2.0],
    [9.0, 1.0],
    [8.0, 2.0]
])

print("Dataset shape:", X.shape)
print(X)
```

#### Dataset Visualization

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    X[:, 0],
    X[:, 1],
    color="steelblue",
    s=90,
    edgecolor="black"
)

plt.title("Original Unclustered Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(alpha=0.3)
plt.show()
```
