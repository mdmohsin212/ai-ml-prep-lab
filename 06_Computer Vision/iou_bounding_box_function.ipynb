````python
import numpy as np
````

```
Here, 

x1, y1 = top-left corner
x2, y2 = bottom-right corner

```

````python
def calculate_iou(box1, box2):
    x1_box1, y1_box1, x2_box1, y2_box1 = box1
    
    x1_box2, y1_box2, x2_box2, y2_box2 = box2
    
    inter_x1 = max(x1_box1, x1_box2)
    inter_y1 = max(y1_box1, y1_box2)
    inter_x2 = min(x2_box1, x2_box2)
    inter_y2 = min(y2_box1, y2_box2)
    
    inter_width = max(0, inter_x2 - inter_x1)
    inter_height = max(0, inter_y2 - inter_y1)
    
    intersection_area = inter_width * inter_height
    
    box1_area = (x2_box1 - x1_box1) * (y2_box1 - y1_box1)
    box2_area = (x2_box2 - x1_box2) * (y2_box2 - y1_box2)
    
    union_area = box1_area + box2_area - intersection_area
    
    if union_area == 0:
        return 0
    
    iou = intersection_area / union_area
    
    return iou
````

````python
box_a = [50, 50, 150, 150]
box_b = [100, 100, 200, 200]

iou_score = calculate_iou(box_a, box_b)

print("IoU Score:", iou_score)
````

````python
box_c = [10, 10, 50, 50]
box_d = [100, 100, 150, 150]

iou_score = calculate_iou(box_c, box_d)

print("IoU Score:", iou_score)
````
