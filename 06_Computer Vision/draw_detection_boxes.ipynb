```python
import cv2
import os
import matplotlib.pyplot as plt
```

```python
detections = [
    {
        "box": [50, 60, 200, 220],
        "label": "person",
        "score": 0.95
    },
    {
        "box": [250, 100, 400, 280],
        "label": "car",
        "score": 0.88
    },
    {
        "box": [120, 300, 300, 450],
        "label": "dog",
        "score": 0.76
    }
]
```

```python
def draw_boxes_and_save(image_path, detections, output_path):
    image = cv2.imread(image_path)
    
    for detections in detections:
        box = detections["box"]
        label = detections["label"]
        score = detections["score"]
        
        x1, y1, x2, y2 = box
        x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
        
        cv2.rectangle(
            image,
            (x1, y1),
            (x2, y2),
            color=(0, 255, 0),
            thickness=2
        )
        text = f"{label} : {score:.2f}"
        
        cv2.rectangle(
            image,
            (x1, y1 - 25),
            (x1 + 150, y1),
            color=(0, 255, 0),
            thickness=-1
        )
        
        cv2.putText(
            image,
            text,
            (x1, y1 - 7),
            cv2.FONT_HERSHEY_SIMPLEX,
            fontScale=0.6,
            color=(0, 0, 0),
            thickness=2
        )
        
        output_dir = os.path.dirname(output_path)
        cv2.imwrite(output_path, image)
        print(f"Output image saved at: {output_path}")
```

```python
input_image_path = "C:\All Programming\Ai Development\Ai-ml-prep\data\man.png"
output_image_path = "C:\All Programming\Ai Development\Ai-ml-prep\data\detected_man.png"

draw_boxes_and_save(
    image_path=input_image_path,
    detections=detections,
    output_path=output_image_path
)
```

```python
output_image = cv2.imread(output_image_path)
output_image_rgb = cv2.cvtColor(output_image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(output_image_rgb)
plt.axis("off")
plt.title("Detection Result")
plt.show()
```
