```python
import re
import numpy as np
import pandas as pd
```

```python
corpus = [
    "I love NLP",
    "I love machine learning",
    "NLP is fun"
]

for i, document in enumerate(corpus):
    print(f"Document {i}: {document}")
```

#### Text Preprocessing Function

```python
def preprocess_text(text):
    text = text.lower()
    tokens = re.findall(r"[a-z]+", text)
    return tokens

for document in corpus:
    print(preprocess_text(document))
```

#### Create Manually Vocabulary

```python
unique_words = set()

for document in corpus:
    tokens = preprocess_text(document)
    
    for word in tokens:
        unique_words.add(word)
        
vocabulary = sorted(unique_words)
print("Vocabulary: ", vocabulary)
print("Vocabulary size:", len(vocabulary))
```

```python
word_to_index = {}

for i, word in enumerate(vocabulary):
    word_to_index[word] = i
    
print("Word to index mapping:")
for word, index in word_to_index.items():
    print(f"{word}: {index}")
```

#### Manual Bag of Words Matrix

```python
bow_matrix = np.zeros(
    (len(corpus), len(vocabulary)),
    dtype=int
)

for document_index, document in enumerate(corpus):
    tokens = preprocess_text(document)
    
    for word in tokens:
        if word in word_to_index:
            word_index = word_to_index[word]
            bow_matrix[document_index][word_index] += 1

print(bow_matrix)
```

```python
bow_dataframe = pd.DataFrame(
    bow_matrix,
    columns=vocabulary,
    index=["Document 1", "Document 2", "Document 3"]
)

display(bow_dataframe)
```

### Complete Manual Bag of Words Class

```python
class ManualBagOfWords:
    def __init__(self):
        self.vocabulary = []
        self.word_to_index = {}
    
    def preprocess(self, text):
        text =  text.lower()
        return re.findall(r"[a-z]", text)
    
    def fit(self, corpus):
        unique_words = set()
        
        for document in corpus:
            tokens = self.preprocess(document)
            for word in tokens:
                unique_words.add(word)
        
        self.vocabulary = sorted(unique_words)
        self.word_to_index = {
            word: index
            for index, word in enumerate(self.vocabulary)
        }
        return self
    
    def transform(self, corpus):
        matrix = np.zeros(
            (len(corpus), len(self.vocabulary)),
            dtype=int
        )

        for document_index, document in enumerate(corpus):
            tokens = self.preprocess(document)

            for word in tokens:
                if word in self.word_to_index:
                    word_index = self.word_to_index[word]
                    matrix[document_index][word_index] += 1

        return matrix

    def fit_transform(self, corpus):    
        self.fit(corpus)
        return self.transform(corpus)

    def get_feature_names(self):
        return self.vocabulary
```

#### Use Manual Vectorizer

```python
vectorizer = ManualBagOfWords()

bow_vectors = vectorizer.fit_transform(corpus)

print("Vocabulary:")
print(vectorizer.get_feature_names())

print("Bag of Words vectors:")
print(bow_vectors)
```

```python
new_documents = [
    "I love fun NLP"
]

new_vectors = vectorizer.transform(new_documents)
new_dataframe = pd.DataFrame(
    new_vectors,
    columns=vectorizer.get_feature_names(),
    index=["New Document"]
)

display(new_dataframe)
```
